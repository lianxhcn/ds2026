# data_visual_setting_codes.ipynb

本 Notebook 用于生成 `data_visual_setting_lec.ipynb` 中使用的全部示例图片和数据。

运行方式：

1. 从上到下依次运行所有代码单元。
2. Notebook 会自动在当前工作路径下创建 `./figs/` 和 `./data/` 子文件夹。
3. 所有图片会统一保存到 `./figs/`。
4. 示例数据会保存到 `./data/`。

图片命名规则：

```text
data_visual_setting_fig##_short_name.png
data_visual_setting_fig##_short_name.svg
```

````python
# ------------------------------------------------------------
# 0. 全局设置：适合 Jupyter Notebook + Quarto 写作
# ------------------------------------------------------------

from pathlib import Path
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import matplotlib.patches as mpatches
from matplotlib.patches import FancyArrowPatch
from matplotlib.lines import Line2D

# 教学演示时可以临时关闭非关键警告
# 正式研究项目中不建议长期忽略所有 warning
warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 0.1 创建项目文件夹
# ------------------------------------------------------------

# 图片统一保存到 ./figs
# 数据统一保存到 ./data
FIG_DIR = Path("./figs")
DATA_DIR = Path("./data")

FIG_DIR.mkdir(parents=True, exist_ok=True)
DATA_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 0.2 中文字体设置
# ------------------------------------------------------------

# 获取当前系统中 matplotlib 能识别的字体名称
available_fonts = {f.name for f in fm.fontManager.ttflist}

# 按顺序寻找可用中文字体
# Windows 通常有 SimHei 或 Microsoft YaHei
# Linux 服务器或 GitHub Actions 环境中，常见选择是 Noto Sans CJK SC
font_candidates = [
    "SimHei",
    "Microsoft YaHei",
    "Noto Sans CJK SC",
    "Noto Sans CJK JP",
    "WenQuanYi Micro Hei",
    "Arial Unicode MS",
]

FONT_FAMILY = next(
    (font for font in font_candidates if font in available_fonts),
    "DejaVu Sans"
)

# 设置默认字体
plt.rcParams["font.family"] = "sans-serif"
plt.rcParams["font.sans-serif"] = [FONT_FAMILY]

# 解决负号显示异常问题
plt.rcParams["axes.unicode_minus"] = False

# ------------------------------------------------------------
# 0.3 全局基础绘图风格
# ------------------------------------------------------------

plt.rcParams.update({
    # Notebook 中的默认显示清晰度
    "figure.dpi": 150,

    # 坐标轴边框：去掉上边框和右边框
    "axes.spines.top": False,
    "axes.spines.right": False,

    # 网格线：开启网格，但透明度较低
    "axes.grid": True,
    "grid.alpha": 0.22,

    # 线条和标记点
    "lines.linewidth": 1.8,
    "lines.markersize": 5,

    # 保存图片时使用白色背景
    "savefig.facecolor": "white",
})

# ------------------------------------------------------------
# 0.4 场景切换函数
# ------------------------------------------------------------

def fig_single():
    """
    单图场景：
    适合时间序列图、分布图、回归拟合图等。
    建议在 Quarto 中使用 width="85%"。
    """
    plt.rcParams.update({
        "figure.figsize": (7.0, 4.2),
        "font.size": 11,
        "axes.labelsize": 11,
        "axes.titlesize": 12,
        "xtick.labelsize": 10,
        "ytick.labelsize": 10,
        "legend.fontsize": 10,
    })


def fig_double():
    """
    双图并排场景：
    适合处理前后对比、两组样本对比、两个模型对比。
    建议在 Quarto 中使用 width="100%"。
    """
    plt.rcParams.update({
        "figure.figsize": (9.0, 3.8),
        "font.size": 10,
        "axes.labelsize": 10,
        "axes.titlesize": 11,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "legend.fontsize": 9,
    })


def fig_triple():
    """
    三图并排场景：
    适合展示流程或三个指标的横向比较。
    建议在 Quarto 中使用 width="100%"，并尽量减少图内文字。
    """
    plt.rcParams.update({
        "figure.figsize": (10.5, 3.2),
        "font.size": 9.5,
        "axes.labelsize": 9.5,
        "axes.titlesize": 10.5,
        "xtick.labelsize": 8.5,
        "ytick.labelsize": 8.5,
        "legend.fontsize": 8.5,
    })


def fig_2x2():
    """
    2 × 2 多子图场景：
    适合模型诊断、多指标比较和稳健性展示。
    建议在 Quarto 中使用 width="95%" 或 width="100%"。
    """
    plt.rcParams.update({
        "figure.figsize": (8.5, 6.5),
        "font.size": 10,
        "axes.labelsize": 10,
        "axes.titlesize": 11,
        "xtick.labelsize": 9,
        "ytick.labelsize": 9,
        "legend.fontsize": 9,
    })

# ------------------------------------------------------------
# 0.5 统一保存函数
# ------------------------------------------------------------

def save_fig(fig, basename, dpi=220, close=True, save_svg=True):
    """
    统一保存图片。

    参数
    ----
    fig : matplotlib.figure.Figure
        需要保存的图形对象。
    basename : str
        不带扩展名的文件名。
    dpi : int
        PNG 图片的导出分辨率。网页和 PDF 兼顾时，建议 200-240。
    close : bool
        是否在保存后关闭图形，批量生成图片时建议为 True。
    save_svg : bool
        是否同时保存 SVG 矢量图。
    """

    # 保存 PNG：兼容性最好，适合网页、推文和普通文档
    fig.savefig(
        FIG_DIR / f"{basename}.png",
        dpi=dpi,
        bbox_inches="tight",
        facecolor="white"
    )

    # 保存 SVG：适合线图、散点图、流程图
    if save_svg:
        fig.savefig(
            FIG_DIR / f"{basename}.svg",
            bbox_inches="tight",
            facecolor="white"
        )

    if close:
        plt.close(fig)

# ------------------------------------------------------------
# 0.6 配色和随机数
# ------------------------------------------------------------

# 项目统一配色。大多数教学图控制在 4-6 个主色即可。
PALETTE = [
    "#1A3A6B",
    "#C8900A",
    "#2E8B57",
    "#CC2222",
    "#7B3F9E",
    "#666666",
]

# 固定随机数生成器，保证模拟数据和图形可复现
RNG = np.random.default_rng(2026)

print(f"当前使用字体: {FONT_FAMILY}")
print(f"图片输出路径: {FIG_DIR.resolve()}")
print(f"数据输出路径: {DATA_DIR.resolve()}")
````

## 1. 生成示例金融数据

````python
# ------------------------------------------------------------
# 1. 生成示例数据：市场收益率与组合收益率
# ------------------------------------------------------------

n = 120
dates = pd.date_range("2016-01-31", periods=n, freq="M")

# 市场收益率：月度收益率，均值略大于 0，波动较高
market_ret = RNG.normal(loc=0.006, scale=0.045, size=n)

# 组合收益率：由 alpha、beta 和随机扰动组成
alpha = 0.002
beta = 1.15
epsilon = RNG.normal(loc=0.000, scale=0.025, size=n)
portfolio_ret = alpha + beta * market_ret + epsilon

# 构造价格指数，便于画时间序列图
market_index = 100 * np.cumprod(1 + market_ret)
portfolio_index = 100 * np.cumprod(1 + portfolio_ret)

df = pd.DataFrame({
    "date": dates,
    "market_ret": market_ret,
    "portfolio_ret": portfolio_ret,
    "market_index": market_index,
    "portfolio_index": portfolio_index,
})

# 保存示例数据，方便后续章节复用
df.to_csv(DATA_DIR / "data_visual_setting_demo_returns.csv", index=False)

df.head()
````

## 2. 图 1：Python 到 Quarto 的显示流程

````python
# ------------------------------------------------------------
# 图 1：Python → Quarto → 输出端的显示流程
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(9.4, 4.6), dpi=150)
ax.set_axis_off()

def add_box(ax, xy, width, height, title, lines, color):
    """
    在图中添加一个圆角矩形说明框。
    """
    x, y = xy
    box = mpatches.FancyBboxPatch(
        (x, y), width, height,
        boxstyle="round,pad=0.02,rounding_size=0.025",
        linewidth=1.4,
        edgecolor=color,
        facecolor=color,
        alpha=0.12
    )
    ax.add_patch(box)

    ax.text(
        x + width / 2, y + height - 0.12,
        title,
        ha="center", va="top",
        fontsize=13, fontweight="bold",
        color=color
    )

    for i, line in enumerate(lines):
        ax.text(
            x + 0.06, y + height - 0.28 - i * 0.12,
            line,
            ha="left", va="top",
            fontsize=10.2,
            color="#222222"
        )

def add_arrow(ax, start, end):
    """
    添加箭头，表示流程方向。
    """
    arrow = FancyArrowPatch(
        start, end,
        arrowstyle="-|>",
        mutation_scale=16,
        linewidth=1.4,
        color="#555555"
    )
    ax.add_patch(arrow)

add_box(
    ax, (0.05, 0.33), 0.25, 0.42,
    "Python 生成图片",
    ["figsize", "dpi", "fontsize", "savefig"],
    PALETTE[0]
)

add_box(
    ax, (0.375, 0.33), 0.25, 0.42,
    "Quarto 插入图片",
    ["width", "caption", "layout", "HTML / PDF"],
    PALETTE[1]
)

add_box(
    ax, (0.70, 0.33), 0.25, 0.42,
    "终端显示图片",
    ["PC", "iPad", "iPhone", "A4 PDF"],
    PALETTE[2]
)

add_arrow(ax, (0.31, 0.54), (0.37, 0.54))
add_arrow(ax, (0.635, 0.54), (0.695, 0.54))

ax.text(
    0.5, 0.15,
    "同一张图的最终效果，由生成层、排版层和显示层共同决定",
    ha="center", va="center",
    fontsize=12.5,
    color="#333333"
)

save_fig(fig, "data_visual_setting_fig01_workflow")
````

## 4. 图 3：同样字号在不同缩放下的视觉差异

````python
# ------------------------------------------------------------
# 图 3：同样 fontsize 在不同显示宽度下的视觉差异
# ------------------------------------------------------------

fig, axes = plt.subplots(1, 3, figsize=(10.5, 3.2), dpi=150, constrained_layout=True)

scenarios = [
    ("显示宽度 100%", 1.00, "视觉字号 ≈ 11.0"),
    ("显示宽度 75%", 0.75, "视觉字号 ≈ 8.3"),
    ("显示宽度 55%", 0.55, "视觉字号 ≈ 6.1"),
]

for ax, (title, scale, note) in zip(axes, scenarios):
    ax.set_axis_off()

    # 模拟图片外框
    rect = mpatches.FancyBboxPatch(
        (0.08, 0.18), 0.84, 0.62,
        boxstyle="round,pad=0.02,rounding_size=0.02",
        linewidth=1.4,
        edgecolor=PALETTE[0],
        facecolor=PALETTE[0],
        alpha=0.08
    )
    ax.add_patch(rect)

    # 用不同实际字号模拟图片缩放后的视觉效果
    ax.text(
        0.50, 0.57,
        "legend fontsize = 11",
        ha="center", va="center",
        fontsize=11 * scale,
        color="#222222"
    )

    ax.text(
        0.50, 0.43,
        "axis label",
        ha="center", va="center",
        fontsize=10 * scale,
        color="#333333"
    )

    ax.text(
        0.50, 0.83,
        title,
        ha="center", va="bottom",
        fontsize=11,
        fontweight="bold",
        color=PALETTE[0]
    )

    ax.text(
        0.50, 0.09,
        note,
        ha="center", va="top",
        fontsize=10,
        color="#555555"
    )

fig.suptitle(
    "同样的 Python 字号，图片被缩放后视觉大小会改变",
    fontsize=13,
    fontweight="bold",
    y=1.04
)
fig.show()
save_fig(fig, "data_visual_setting_fig03_font_scaling")
````

## 5. 图 4：单图、双图和三图布局比较

````python
# ------------------------------------------------------------
# 图 4：单图、双图、三图布局比较
# ------------------------------------------------------------

fig, axes = plt.subplots(3, 1, figsize=(9.0, 6.2), dpi=150, constrained_layout=True)

layouts = [
    ("单图：figsize=(7.0, 4.2)，Quarto width=85%", 1, PALETTE[0]),
    ("双图：figsize=(9.0, 3.8)，Quarto width=100%", 2, PALETTE[1]),
    ("三图：figsize=(10.5, 3.2)，Quarto width=100%", 3, PALETTE[2]),
]

for ax, (title, ncols, color) in zip(axes, layouts):
    ax.set_axis_off()
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)

    ax.text(
        0.02, 0.88,
        title,
        ha="left", va="center",
        fontsize=11,
        fontweight="bold",
        color=color
    )

    margin = 0.04
    gap = 0.035
    width = (1 - 2 * margin - (ncols - 1) * gap) / ncols

    for j in range(ncols):
        x0 = margin + j * (width + gap)
        rect = mpatches.FancyBboxPatch(
            (x0, 0.18), width, 0.52,
            boxstyle="round,pad=0.01,rounding_size=0.02",
            linewidth=1.3,
            edgecolor=color,
            facecolor=color,
            alpha=0.10
        )
        ax.add_patch(rect)

        ax.text(
            x0 + width / 2, 0.44,
            f"子图 {j + 1}",
            ha="center", va="center",
            fontsize=10 if ncols < 3 else 9,
            color="#222222"
        )

save_fig(fig, "data_visual_setting_fig04_single_multi_compare")
````

## 6. 图 5：单图标准模板

````python
# ------------------------------------------------------------
# 图 5：单图模板
# ------------------------------------------------------------

fig_single()

fig, ax = plt.subplots()

ax.plot(
    df["date"],
    df["portfolio_index"],
    label="组合净值",
    color=PALETTE[0]
)

ax.plot(
    df["date"],
    df["market_index"],
    label="市场指数",
    color=PALETTE[1],
    linestyle="--"
)

ax.set_xlabel("日期")
ax.set_ylabel("指数，初始值 = 100")
ax.legend(frameon=True)

# 单图一般不需要图内大标题，标题和解释交给 Quarto caption
save_fig(fig, "data_visual_setting_fig05_single_plot")
````

## 7. 图 6：双图并排模板

````python
# ------------------------------------------------------------
# 图 6：双图并排模板
# ------------------------------------------------------------

fig_double()

fig, axes = plt.subplots(1, 2, constrained_layout=True)

# 左图：收益率分布
ax = axes[0]
ax.hist(
    df["portfolio_ret"] * 100,
    bins=20,
    color=PALETTE[0],
    alpha=0.78,
    edgecolor="white"
)
ax.axvline(
    df["portfolio_ret"].mean() * 100,
    color=PALETTE[3],
    linewidth=1.6,
    linestyle="--",
    label="均值"
)
ax.set_title("(a) 组合收益率分布")
ax.set_xlabel("月度收益率 (%)")
ax.set_ylabel("频数")
ax.legend(frameon=True)

# 右图：CAPM 风格散点拟合
ax = axes[1]
x = df["market_ret"].to_numpy() * 100
y = df["portfolio_ret"].to_numpy() * 100

coef = np.polyfit(x, y, deg=1)
x_grid = np.linspace(x.min(), x.max(), 100)
y_fit = coef[0] * x_grid + coef[1]

ax.scatter(
    x, y,
    s=24,
    color=PALETTE[1],
    alpha=0.72,
    edgecolor="white",
    linewidth=0.4,
    label="样本点"
)
ax.plot(
    x_grid, y_fit,
    color=PALETTE[0],
    linewidth=1.8,
    label="线性拟合"
)
ax.set_title("(b) 市场收益率与组合收益率")
ax.set_xlabel("市场收益率 (%)")
ax.set_ylabel("组合收益率 (%)")
ax.legend(frameon=True)

save_fig(fig, "data_visual_setting_fig06_two_panel")
````

## 8. 图 7：三图并排模板

````python
# ------------------------------------------------------------
# 图 7：三图并排模板
# ------------------------------------------------------------

fig_triple()

fig, axes = plt.subplots(1, 3, constrained_layout=True)

# 计算滚动均值和回撤
df_plot = df.copy()
df_plot["portfolio_ma6"] = df_plot["portfolio_ret"].rolling(6).mean()
df_plot["portfolio_cummax"] = df_plot["portfolio_index"].cummax()
df_plot["drawdown"] = df_plot["portfolio_index"] / df_plot["portfolio_cummax"] - 1

# 子图 1：月度收益率
ax = axes[0]
ax.plot(
    df_plot["date"],
    df_plot["portfolio_ret"] * 100,
    color=PALETTE[0],
    linewidth=1.4
)
ax.axhline(0, color="#777777", linewidth=1.0)
ax.set_title("(a) 月度收益率")
ax.set_xlabel("日期")
ax.set_ylabel("%")

# 子图 2：滚动均值
ax = axes[1]
ax.plot(
    df_plot["date"],
    df_plot["portfolio_ma6"] * 100,
    color=PALETTE[1],
    linewidth=1.4
)
ax.axhline(0, color="#777777", linewidth=1.0)
ax.set_title("(b) 6 期滚动均值")
ax.set_xlabel("日期")
ax.set_ylabel("%")

# 子图 3：回撤
ax = axes[2]
ax.fill_between(
    df_plot["date"],
    df_plot["drawdown"] * 100,
    0,
    color=PALETTE[3],
    alpha=0.22
)
ax.plot(
    df_plot["date"],
    df_plot["drawdown"] * 100,
    color=PALETTE[3],
    linewidth=1.3
)
ax.set_title("(c) 回撤")
ax.set_xlabel("日期")
ax.set_ylabel("%")

# 三图并排时，减少横坐标标签数量，避免拥挤
for ax in axes:
    ax.tick_params(axis="x", rotation=25)

save_fig(fig, "data_visual_setting_fig07_three_panel")
````

## 9. 图 8：2 × 2 多子图模板

````python
# ------------------------------------------------------------
# 图 8：2 × 2 多子图模板
# ------------------------------------------------------------

fig_2x2()

fig, axes = plt.subplots(2, 2, constrained_layout=True)

# (a) 时间序列
ax = axes[0, 0]
ax.plot(
    df["date"],
    df["portfolio_index"],
    color=PALETTE[0],
    label="组合"
)
ax.plot(
    df["date"],
    df["market_index"],
    color=PALETTE[1],
    linestyle="--",
    label="市场"
)
ax.set_title("(a) 净值走势")
ax.set_xlabel("日期")
ax.set_ylabel("指数")
ax.legend(frameon=True)

# (b) 收益率分布
ax = axes[0, 1]
ax.hist(
    df["portfolio_ret"] * 100,
    bins=20,
    color=PALETTE[0],
    alpha=0.75,
    edgecolor="white"
)
ax.set_title("(b) 收益率分布")
ax.set_xlabel("月度收益率 (%)")
ax.set_ylabel("频数")

# (c) 散点拟合
ax = axes[1, 0]
x = df["market_ret"].to_numpy() * 100
y = df["portfolio_ret"].to_numpy() * 100
coef = np.polyfit(x, y, deg=1)
x_grid = np.linspace(x.min(), x.max(), 100)
y_fit = coef[0] * x_grid + coef[1]

ax.scatter(
    x, y,
    s=22,
    color=PALETTE[1],
    alpha=0.72,
    edgecolor="white",
    linewidth=0.4
)
ax.plot(
    x_grid,
    y_fit,
    color=PALETTE[0],
    linewidth=1.8
)
ax.set_title("(c) 市场暴露")
ax.set_xlabel("市场收益率 (%)")
ax.set_ylabel("组合收益率 (%)")

# (d) 拟合残差
ax = axes[1, 1]
resid = y - (coef[0] * x + coef[1])

ax.scatter(
    np.arange(len(resid)),
    resid,
    s=20,
    color=PALETTE[2],
    alpha=0.72,
    edgecolor="white",
    linewidth=0.4
)
ax.axhline(0, color="#777777", linewidth=1.0)
ax.set_title("(d) 拟合残差")
ax.set_xlabel("样本序号")
ax.set_ylabel("残差")

for ax in axes.ravel():
    ax.tick_params(axis="x", rotation=20)

save_fig(fig, "data_visual_setting_fig08_four_panel")
````
